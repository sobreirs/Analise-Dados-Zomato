```python
import pandas as pd

df = pd.read_csv("../data/raw/Zomato_Dataset.csv")
```

```python
# Verificando se há valores nulos nas colunas

df.isnull().sum()
```

```python
# Verificando se há duplicatas no Dataset

print(df.duplicated().sum())
```

```python
# Verificando se há duplicatas nos ID

print(df["ID"].duplicated().any())
```

```python
# Verificando se a idade do entregador está de acordo

print(f'Tem idades negativas?', (df["Delivery_person_Age"] < 0).any())

print(f'Tem idade menor que 18?', (df["Delivery_person_Age"] < 18).any())
print(f'Tem idade maior que 60?', (df["Delivery_person_Age"] > 60).any())

print(f'Quantos motoristas tem menos de 18 anos?', (df["Delivery_person_Age"] < 18).sum())
```

```python
# Verificando as notas dos entregadores

print(f'Menor nota:', df["Delivery_person_Ratings"].min())
print(f'Maior nota:', df["Delivery_person_Ratings"].max()) 

# Nota acima de 5 = inválida

print(f'Quantas notas maiores que 5?', (df["Delivery_person_Ratings"] > 5).sum())
```

```python
# Tratando a latitude e longitude

# Quantos têm sinal negativo (quando deveriam ser positivos, já que é Índia)
print("Restaurant_lat negativa:", (df["Restaurant_latitude"] < 0).sum())
print("Restaurant_lon negativa:", (df["Restaurant_longitude"] < 0).sum())
print("Delivery_lat negativa:", (df["Delivery_location_latitude"] < 0).sum())
print("Delivery_lon negativa:", (df["Delivery_location_longitude"] < 0).sum())

# Quantos são exatamente (0,0)
null_island_rest = (df["Restaurant_latitude"] == 0) & (df["Restaurant_longitude"] == 0)
null_island_cli = (df["Delivery_location_latitude"] == 0) & (df["Delivery_location_longitude"] == 0)
print("Null Island restaurante:", null_island_rest.sum())
print("Null Island cliente:", null_island_cli.sum())
```
